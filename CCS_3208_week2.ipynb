### Experiment with different delays and echo strengths

````python
delays_to_test = [0.10, 0.50, 0.75] # seconds
echo_strengths_to_test = [0.2, 0.4, 0.7]

# Re-initialize the original signal (x) to ensure consistent starting point
duration = 2 # seconds
t = np.arange(int(fs * duration)) / fs # Time vector for 2 seconds
x = 0.5 * np.sin(2 * np.pi * 440 * t) # Original 440 Hz sine wave

for delay_s in delays_to_test:
    for strength in echo_strengths_to_test:
        delay_samp = int(delay_s * fs)

        # Create a copy of the original signal for each experiment
        echo_signal = x.copy()

        # Apply echo effect, ensuring we don't go out of bounds
        # The echo portion starts at `delay_samp` and extends to the end of the signal
        # The original signal `x` is truncated to match the length of the echo portion
        if delay_samp < len(echo_signal):
            echo_signal[delay_samp:] += strength * x[:-delay_samp]
        else:
            # Handle cases where delay_samp is too large for the signal length
            # (e.g., if duration was too short or delay_s too long)
            print(f"Warning: Delay {delay_s}s is too long for a signal of length {len(x)} samples. Skipping this combination.")
            continue

        plt.figure(figsize=(12, 4))
        plt.plot(t[:1000], echo_signal[:1000]) # Plot first 1000 samples for visualization
        plt.xlabel("Time (s)")
        plt.ylabel("Amplitude")
        plt.title(f"Echo: Delay = {delay_s:.2f}s, Strength = {strength:.1f}")
        plt.grid(True)
        plt.show()
````

### Effect of Increasing Delay and Echo Strength

*   **Increasing Delay:** A longer delay time between the original signal and its echo will result in a more distinct and separate echo sound. If the delay is very short, the echo might just sound like a thickening or reverberation of the original sound. As the delay increases, the echo becomes more discernible as a distinct repeat of the original sound, rather than blending with it.

*   **Increasing Echo Strength (Gain):** A higher echo strength (or gain) means the echo signal is louder relative to the original signal. This makes the echo more prominent and noticeable. If the strength is too high, the echo can become as loud or even louder than the original sound, potentially leading to feedback or an unnatural, overwhelming effect.

### Mathematical Expression for a 0.25-second, 40% Echo at 44.1 kHz

Given:
*   Original signal: $x(t)$
*   Sampling frequency: $f_s = 44.1 \text{ kHz}$ (or $44100 \text{ Hz}$)
*   Delay: $\tau = 0.25 \text{ s}$
*   Echo strength (gain): $g = 0.4$

First, calculate the delay in samples:
$D = \tau \cdot f_s = 0.25 \text{ s} \cdot 44100 \text{ Hz} = 11025 \text{ samples}$

The echoed signal, $y[n]$, at a discrete time index $n$ can be expressed as the sum of the original signal and its delayed, attenuated version:

$y[n] = x[n] + g \cdot x[n - D]$

Substituting the given values:

$y[n] = x[n] + 0.4 \cdot x[n - 11025]$

### Comparison of Fine Details and Edges Before and After Sharpening

**Before Sharpening (Original Grayscale - `gray_cv`):**
*   Details and edges might appear softer, less defined, or slightly blurred.
*   Transitions between different intensity regions are gradual.

**After Sharpening (`sharpened`):**
*   Edges, such as outlines of objects, text, or high-contrast areas, become much more pronounced and crisp.
*   Fine textures and small details that were subtle in the original image are now more visible.
*   The overall visual effect is an image that looks clearer and more focused. This is because the sharpening kernel enhances the difference between a pixel and its neighbors, effectively increasing local contrast at edges.

However, it's worth noting that aggressive sharpening can sometimes introduce artifacts or make the image look unnaturally harsh if overdone.

### Manual Calculation of Sharpening Kernel Output

Given the sharpening kernel:
`kernel = np.array([[0, -1, 0], [-1, 5, -1], [0, -1, 0]])`

And a neighborhood with center value 20 and four direct neighbors of value 10:
```
  10
10 20 10
  10
```

Let's represent the neighborhood as a 3x3 matrix where unknown values are 0 (assuming they don't contribute to direct neighbors):
```
  0  10  0
 10  20 10
  0  10  0
```

The convolution operation for the center pixel is calculated by multiplying each element of the kernel with the corresponding element in the image neighborhood and summing the results:

Output Pixel Value = (0 * 0) + (-1 * 10) + (0 * 0) +
                     (-1 * 10) + (5 * 20) + (-1 * 10) +
                     (0 * 0) + (-1 * 10) + (0 * 0)

Output Pixel Value = 0 - 10 + 0 -
                     10 + 100 - 10 +
                     0 - 10 + 0

Output Pixel Value = -10 - 10 + 100 - 10 - 10
Output Pixel Value = 100 - 40
Output Pixel Value = 60

So, the output of the sharpening kernel for the center pixel with value 20, surrounded by four direct neighbors of value 10, would be **60**.

### Explanation of why sharpening can also make noise more visible

Sharpening filters work by enhancing the contrast of edges in an image. They do this by detecting sudden changes in pixel intensity (which typically occur at edges) and amplifying those differences. The kernel used, like the one here, has positive values at the center and negative values for immediate neighbors, which effectively highlights the difference between a pixel and its surroundings.

**The problem with noise is that it often manifests as small, abrupt, random variations in pixel intensity.** These variations can be mistaken for fine details or weak edges by a sharpening filter. When the filter processes these noisy regions, it treats the small intensity fluctuations as if they were legitimate edges and amplifies them.

Therefore, if an image contains noise, applying a sharpening filter will not only enhance real edges but will also **amplify the noise**, making the noisy areas appear more prominent, grainy, or speckled. This can degrade the overall quality of the image, as the enhanced noise becomes more noticeable and distracting.

### Explanation of Frame-by-Frame Video Processing

This activity demonstrates frame-by-frame video processing by:

1.  **Iterating through frames:** The `while True: ok, frame = cap.read()` loop reads the video one frame at a time until no more frames are available.
2.  **Individual frame access:** Each `frame` variable holds a single image (a NumPy array) representing that specific moment in the video.
3.  **Independent processing:** Operations like `cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)` are applied to each `frame` independently. This means the processing of one frame does not directly depend on the content of the previous or next frames.
4.  **Saving processed frames:** The `cv2.imwrite()` function saves each processed grayscale frame as a separate image file, effectively extracting still images from the video stream.

This approach is fundamental for many video analysis tasks where transformations or analyses are performed on individual snapshots of the video content.

### Maximum Time Available Per Frame for Real-Time Processing (30 FPS)

If a video runs at 30 frames per second (FPS), it means 30 frames are displayed every second. To achieve real-time processing, the system must process each frame within the time allocated for that frame.

Time per frame = 1 second / 30 frames = 0.0333... seconds

So, the maximum time available per frame for real-time processing is approximately **33.3 milliseconds**.

### Examples of Temporal Processing

Temporal processing in video requires information from multiple frames to analyze changes or patterns over time, rather than just isolated frames. Here are two examples:

1.  **Object Tracking:** To track an object (e.g., a car, a person) moving across multiple frames, the system needs to identify the object in one frame and then predict its position in subsequent frames. This involves correlating visual features of the object across a sequence of frames to maintain its identity and trajectory. A single frame only shows the object's presence, not its movement.
2.  **Motion Detection/Estimation:** Detecting movement or estimating optical flow (the apparent motion of objects, surfaces, and edges in a visual scene) requires comparing pixel intensities or features between consecutive frames. If a region of pixels changes significantly between Frame N and Frame N+1, it indicates motion. Without looking at multiple frames, it's impossible to discern movement; only static patterns would be visible.

Activity F3: Simple ADSR Envelope

````python
fs = 44100
duration = 1.0
t = np.arange(int(fs * duration)) / fs
tone = 0.5 * np.sin(2*np.pi*440*t)

attack = int(0.10 * fs)
decay = int(0.15 * fs)
release = int(0.20 * fs)
sustain_end = len(tone) - release

env = np.ones(len(tone))
env[:attack] = np.linspace(0, 1, attack)
env[attack:attack+decay] = np.linspace(1, 0.7, decay)
env[attack+decay:sustain_end] = 0.7
env[sustain_end:] = np.linspace(0.7, 0, release)

shaped = tone * env

plt.plot(t, env)
plt.xlabel("Time (s)")
plt.ylabel("Envelope Level")
plt.title("ADSR-style Amplitude Envelope")
plt.grid(True)
plt.show()
````
